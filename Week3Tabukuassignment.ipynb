```python
## Task 1: Study Time vs. Exam Score

import matplotlib.pyplot as plt

# Given Data
study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]

# Create the scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(study_hours, exam_scores, color='purple', marker='o', s=100, edgecolor='black')

# Add title and labels
plt.title('Study Time vs. Exam Score', fontsize=14, fontweight='bold')
plt.xlabel('Study Hours', fontsize=12)
plt.ylabel('Exam Score', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

As study hours increase, exam scores also increase in a strong positive linear relationship. This suggests that spending more time studying tends to result in higher exam scores.

```python
### Task 2: Temperature Over One Week

# Given Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
city_a = [68, 70, 72, 71, 74, 76, 75]
city_b = [62, 64, 67, 66, 69, 71, 70]

# Create the line plot
plt.figure(figsize=(9, 5))
plt.plot(days, city_a, label='City A', color='teal', marker='o', linestyle='-', linewidth=2)
plt.plot(days, city_b, label='City B', color='coral', marker='s', linestyle='--', linewidth=2)

# Add title, labels, and legend
plt.title('Weekly Temperatures for City A and City B', fontsize=14, fontweight='bold')
plt.xlabel('Days of the Week', fontsize=12)
plt.ylabel('Temperature (°F)', fontsize=12)
plt.legend()

# Add gridlines
plt.grid(True, linestyle=':', alpha=0.7)

plt.show()
```

City A was generally warmer during the week, as its temperature line stays consistently higher than City B's across all days.

```python
## Task 3A: Basic Bar Chart for Crime Rates in 2023

import pandas as pd

# Create DataFrame from the crime dataset
crime_data = {
    'Offense': ['Aggravated Assault', 'Burglary', 'Larceny', 'Motor Vehicle Theft', 'Murder', 'Rape', 'Robbery'],
    'Rate_2023': [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5],
    'Change': [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]
}
df_crime = pd.DataFrame(crime_data)

# Create the bar chart
plt.figure(figsize=(10, 6))
plt.bar(df_crime['Offense'], df_crime['Rate_2023'], color='royalblue', edgecolor='black')

# Add labels, title, and formatting
plt.title('Offenses per 100,000 People (2023)', fontsize=14, fontweight='bold')
plt.xlabel('Offense Type', fontsize=12)
plt.ylabel('Rate per 100,000 People', fontsize=12)
plt.xticks(rotation=30, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
## Task 3B: Change in Crime Rates from 2022 to 2023

# Assign unique colors
colors = ['crimson' if val < 0 else 'forestgreen' for val in df_crime['Change']]

plt.figure(figsize=(10, 6))
plt.barh(df_crime['Offense'], df_crime['Change'], color=colors, edgecolor='black')

plt.title('Change in Offense Rates per 100,000 People (2022–2023)', fontsize=14, fontweight='bold')
plt.xlabel('Percentage Change (%)', fontsize=12)
plt.ylabel('Offense Type', fontsize=12)
plt.axvline(0, color='black', linewidth=1) # Reference line at 0
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

Task 3B is more useful for understanding trajectory and relative trends because it explicitly highlights whether a crime type is rising or falling year-over-year. Task 3A only shows the absolute magnitude of a crime rate in 2023, making it hard to see which direction crime rates are heading.

```python
## Task 3C: Improve the Clarity of Your Bar Graph

# Sort the DataFrame by 'Change' to make the chart easier to read
df_sorted = df_crime.sort_values('Change', ascending=True)
sorted_colors = ['crimson' if val < 0 else 'forestgreen' for val in df_sorted['Change']]

plt.figure(figsize=(11, 6))
bars = plt.barh(df_sorted['Offense'], df_sorted['Change'], color=sorted_colors, edgecolor='black')

# Add data value labels directly on the bars
for bar in bars:
    width = bar.get_width()
    label_x_pos = width + (0.5 if width >= 0 else -1.5)
    plt.text(label_x_pos, bar.get_y() + bar.get_height()/2, f'{width:+.1f}%',
             va='center', ha='left' if width >= 0 else 'right', fontsize=10, fontweight='bold')

plt.title('Sorted Percentage Change in Offense Rates (2022–2023)', fontsize=14, fontweight='bold')
plt.xlabel('Percentage Change (%)', fontsize=12)
plt.ylabel('Offense Type', fontsize=12)
plt.axvline(0, color='black', linewidth=1)
plt.xlim(-16, 16) # Add breathing room for text labels
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
## Task 3D: Multi-Series Grouped Bar Chart

import numpy as np
import matplotlib.pyplot as plt

# Given Data
companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

x = np.arange(len(companies))  # the label locations
width = 0.25  # the width of the bars

plt.figure(figsize=(10, 6))

# Plot bars side by side using blue, red, and black
plt.bar(x - width, product1, width, label='Product 1', color='blue', edgecolor='black')
plt.bar(x, product2, width, label='Product 2', color='red', edgecolor='black')
plt.bar(x + width, product3, width, label='Product 3', color='black', edgecolor='white') # white edge so black bar stands out

# Add labels, title, and custom x-axis tick labels
plt.title('Product Sales Comparison Across Companies', fontsize=14, fontweight='bold')
plt.xlabel('Companies', fontsize=12)
plt.ylabel('Sales Volume', fontsize=12)
plt.xticks(x, companies)
plt.legend()

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
